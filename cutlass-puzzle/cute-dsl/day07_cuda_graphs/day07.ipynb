# Day 07 - CUDA Graph Capture and Replay

This challenge follows NVIDIA's [CUDA Graph notebook](../notebooks/11_cuda_graphs.ipynb). It captures repeated launches of a stateful increment kernel and verifies replay numerically.


## Required order

```text
PyTorch current stream -> CUstream
cute.compile(...)      -> compiled callable
torch.cuda.graph(...)  -> enqueue compiled callable only
graph.replay()         -> one host submission
```

JIT compilation is deliberately outside capture because module loading and allocation are not capture-safe.

## Connection to GEMM
