## The slot protocol

| Producer | Consumer |
| --- | --- |
| `acquire_and_advance()` waits for an empty slot | `wait_and_advance()` waits for a full slot |
| write `stage_buffer[handle.index]` | read the same slot |
| `handle.commit()` publishes it | `handle.release()` makes it reusable |

With `S` stages, iteration `i` reuses slot `i % S`. Barrier phase bits distinguish the current trip around the ring from the previous one.
