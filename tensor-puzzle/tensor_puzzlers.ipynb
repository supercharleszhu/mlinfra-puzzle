## Rules

1. Solve each puzzle with one expression shorter than 80 columns.
2. You may use `@`, arithmetic, comparison, `shape`, indexing, and earlier puzzle functions.
3. Use `arange` instead of a Python loop and `where` instead of a Python `if`.
4. Do not call convenience implementations such as `view`, `sum`, `take`, `squeeze`, or a constructor that directly creates the requested result.

Broadcasting aligns dimensions from the right. A dimension is compatible when the sizes match or one side has size one. Insert `None` to create a size-one dimension.


```python
def arange(i: int):
    """Use this helper to replace a Python loop."""
    return torch.tensor(range(i))


[arange(i) for i in (5, 3, 9)]
```

```python
# Broadcasting examples.
broadcast_examples = [
    (arange(4), arange(5)[:, None]),
    (arange(3)[:, None], arange(2)),
]
[(a.shape, b.shape, (a + b).shape) for a, b in broadcast_examples]
```

```python
def where(q, a, b):
    """Use this helper to replace a Python if-statement."""
    return (q * a) + (~q) * b


where(
    tensor([[False, True], [True, False]]),
    tensor([[0], [10]]),
    tensor([-10, 0]),
)
```

## Puzzle 14 - sequence_mask


Compute [sequence_mask](https://www.tensorflow.org/api_docs/python/tf/sequence_mask) - pad out to length per batch.

```python
def sequence_mask_spec(values, length, out):
    for i in range(len(out)):
        for j in range(len(out[0])):
            if j < length[i]:
                out[i][j] = values[i][j]
            else:
                out[i][j] = 0
    
def sequence_mask(values: TT["i", "j"], length: TT["i", int]) -> TT["i", "j"]:
    # fill in
    raise NotImplementedError


def constraint_set_length(d):
    d["length"] = d["length"] % d["values"].shape[1]
    return d


test_sequence = make_test("sequence_mask",
    sequence_mask, sequence_mask_spec, constraint=constraint_set_length
)
```

## Puzzle 21 - bucketize

Compute [`torch.bucketize`](https://pytorch.org/docs/stable/generated/torch.bucketize.html): for every value, return the index of the interval selected by sorted boundaries.


```python
def bucketize_spec(v, boundaries, out):
    for i, val in enumerate(v):
        out[i] = 0
        for j in range(len(boundaries)-1):
            if val >= boundaries[j]:
                out[i] = j + 1
        if val >= boundaries[-1]:
            out[i] = len(boundaries)


def constraint_set(d):
    d["boundaries"] = np.abs(d["boundaries"]).cumsum()
    return d

            
def bucketize(v: TT["i"], boundaries: TT["j"]) -> TT["i"]:
    # fill in
    raise NotImplementedError

test_bucketize = make_test("bucketize", bucketize, bucketize_spec,
                           constraint=constraint_set)
```
